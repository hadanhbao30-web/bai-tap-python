```python
#1 . Tạo dataframe
import pandas as df

 = pd.DataFrame({"Họ và tên": ["An", "Bình", "Cường", "Dung", "Bảo", "Hương", "Khoa", "Lan", "Minh", "Ngọc"], "Chuyên cần": [9,8,10,7,9,8,10,9,7,8],"Giữa kỳ": [8,7,9,6,8,7,9,8,6,7],"Cuối kỳ": [9,8,10,7,9,8,9,9,7,8]})
display(df)
```

```python
#2. Tính điểm tổng kết
df["Tong_ket"] = df["Chuyên cần"]*0.2 + df["Giữa kỳ"]*0.3 + df["Cuối kỳ"]*0.5

display(df)
```

```python
#3 .Xếp loại sinh viên
def xep_loai(diem):
    if diem >= 8.5:
       return("giỏi")
    elif 7<= diem <8.5:
       return("khá")
    elif 5<= diem <7:
       return("trung bình")
    else :
      return("yếu")

df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)
display(df)
```

```python
#4. Thống kê
print("Điểm trung bình:", df["Tong_ket"].mean())
print(df.loc[df["Tong_ket"].idxmax()])
print(df.loc[df["Tong_ket"].idxmin()])
print("Số sinh viên đạt:", (df["Tong_ket"] >= 5).sum())
```

```python
#5. Trực quan hóa
import matplotlib.pyplot as plt
plt.bar(df["Họ và tên"], df["Tong_ket"])
plt.title("Điểm tổng kết của 10 sinh viên")
plt.xlabel("Tên sinh viên")
plt.ylabel("Điểm tổng kết")
plt.xticks(rotation=45)
plt.ylim(0, 10)
plt.tight_layout()
plt.show()
```
